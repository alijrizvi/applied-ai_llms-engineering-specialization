````python
%%writefile utils.py
# Our Custom Helper Functions

# utils.py

from openai import OpenAI
from IPython.display import Markdown, display # Required for logging functions
import json # Used for pretty-printing tool arguments if they are complex

def generate_insights(text, client):

    prompt = f"""
    You are a football analytics expert.

    Analyze the following text and identify:

    1. Surprising trends
    2. Hidden statistical insights
    3. Underappreciated teams or players
    4. Tactical patterns

    Text:
    {text}
    """

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role":"user","content":prompt}
        ]
    )

    return response.choices[0].message.content

# --- Logging and Display Functions ---
def log_agent_title_html(title, icon):
    display(Markdown(f"## {icon} {title}"))

def log_final_summary_html(summary):
    display(Markdown(f"### Final Summary\n{summary}"))

def log_tool_call_html(tool_name, tool_args):
    try:
        # Attempt to pretty print JSON args, otherwise print as string
        display(Markdown(f"**Calling Tool:** `{tool_name}` with arguments: `{json.dumps(tool_args, indent=2)}`"))
    except (TypeError, json.JSONDecodeError):
        display(Markdown(f"**Calling Tool:** `{tool_name}` with arguments: `{tool_args}`"))

def log_tool_result_html(result):
    display(Markdown(f"**Tool Result:** `{result}`"))

def log_unexpected_html():
    display(Markdown("⚠️ Unexpected: No tool_calls or content returned"))
````

### 4.1: League Research Agent

The first agent/step. You will see what it does, here:

````python
def league_research_agent(return_messages: bool = False):

    utils.log_agent_title_html("League Research Agent", "🕵️‍♂️")

    prompt_ = f"""
You are a football leagues' research agent tasked with preparing a \
trend analysis for each league's season for all metrics.

Your goal:
1. Explore the current tactical trends in Europe's top five football leagues using web search.
2. Review the historical trends of each league.
3. Compare and Review the similarities and differences in both kinds of trends.
4. Suggest future shifts; the next possible tactical trends and surprising findings.

You can call the following tools:
- get_article_text: to discover the latest news articles and reports.
- get_premier_league_insights: to inspect the past (season 2023-2024) tactical and metrical trends in England.
- get_la_liga_insights: to inspect the past (season 2023-2024) tactical and metrical trends in Spain.
- get_serie_a_insights: to inspect the past (season 2023-2024) tactical and metrical trends in Italy.
- get_bundesliga_insights: to inspect the past (season 2023-2024) tactical and metrical trends in Germany.
- get_portugal_liga_insights: to inspect the past (season 2023-2024) tactical and metrical trends in France.

Once your analysis is complete, summarize:
- The top past 2–3 trends you found.
- Where you believe football's current tactical trends lie.
- Future predictions along with justification of why you believe they will come about.
"""
    messages = [{"role": "user", "content": prompt_}]
    tools_ = tools.get_available_tools()

    while True:
        response = client.chat.completions.create(
            model="gpt-4o-mini", # Changed model name from "openai:o4-mini" to "gpt-4o-mini"
            messages=messages,
            tools=tools_,
            tool_choice="auto"
        )

        msg = response.choices[0].message

        if msg.content:
            utils.log_final_summary_html(msg.content)
            return (msg.content, messages) if return_messages else msg.content

        if msg.tool_calls:

            # Add assistant message ONCE
            messages.append(msg)

            for tool_call in msg.tool_calls:

                utils.log_tool_call_html(
                    tool_call.function.name,
                    tool_call.function.arguments
                )

                # Pass the current global scope (from the notebook) to handle_tool_call
                result = tools.handle_tool_call(tool_call, globals())

                utils.log_tool_result_html(result)

                messages.append(
                    tools.create_tool_response_message(
                        tool_call,
                        result
                    )
                )

            continue
````

### 4.2: Infographic Designer Agent

````python
import base64

def infographic_designer_agent(trend_insights: str, caption_style: str = "short punchy", size: str = "1024x1024") -> dict:
    utils.log_agent_title_html("Graphic Designer Agent", "🎨")

    # Step 1: Generate prompt and caption using aisuite
    system_message = (
        "You are a visual marketing assistant focused on conveying infographics to fans to convey unexpected findings for their \
        teams. Based on the input trend insights, \
        write a creative and visual prompt for an AI image generation model, and also a short caption."
    )
    user_prompt = f"""
Trend insights:
{trend_insights}
Please output:
1. A vivid, descriptive prompt to guide image generation.
2. A fun and insightful caption in style, for each image: {caption_style}.
Respond ONLY with the JSON object in this format:
{{"prompt": "...", "caption": "..."}}
Do not include any other text, markdown formatting (like ```json), or conversational elements.
"""
    chat_response = client.chat.completions.create(
        model = "gpt-4o-mini",
        messages = [
            {"role": "system", "content": system_message},
            {"role": "user", "content": user_prompt}
        ]
    )
    content = chat_response.choices[0].message.content.strip()

    parsed = {}
    json_start = content.find('{')
    if json_start != -1:
        json_potential_string = content[json_start:]
        decoder = json.JSONDecoder()
        try:
            parsed, end_idx = decoder.raw_decode(json_potential_string)
            # Optionally, log if there's remaining data after the first JSON object
            if json_potential_string[end_idx:].strip():
                print(f"Warning: Model outputted extra data after the primary JSON. Remainder: '{json_potential_string[end_idx:].strip()}'")
        except json.JSONDecodeError as e:
            parsed = {"error": f"Failed to parse JSON from response: {e}", "raw": content, "extracted_potential_string": json_potential_string}
    else:
        parsed = {"error": "No JSON object start found in response", "raw": content}

    prompt = parsed.get("prompt")
    caption = parsed.get("caption")

    if not prompt or not caption:
        raise ValueError(f"Could not extract prompt or caption from model response. Parsed content: {parsed}. Raw content: {content}")

    # Step 2: Generate Image
    openai_client = openai.OpenAI(api_key = OPENAI_API_KEY)
    image_response = openai_client.images.generate(
    model = "gpt-image-1",
        prompt = prompt,
        size = size,
        quality = "high",
        n = 1,
        # no response_format — gpt-image-1 always returns b64_json
    )

    b64 = image_response.data[0].b64_json
    img_bytes = base64.b64decode(b64)
    img = Image.open(BytesIO(img_bytes))

    image_path = "generated_image.png"
    img.save(image_path)

    utils.log_final_summary_html(f"""
        <h3>Generated Image and Caption</h3>
        <p><strong>Image Path:</strong> <code>{image_path}</code></p>
        <p><strong>Generated Image:</strong></p>
        <img src="{image_path}" alt="Generated Image" style="max-width: 100%; height: auto; border: 1px solid #ccc; border-radius: 8px; margin-top: 10px; margin-bottom: 10px;">
        <p><strong>Prompt:</strong> {prompt}</p>
    """)

    return {
        "image_path": image_path,
        "prompt": prompt,
        "caption": caption,
    }
````
