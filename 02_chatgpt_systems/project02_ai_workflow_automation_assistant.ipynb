## Importing Libraries & Initializing the OpenAI Client

```
from openai import OpenAI
from google.colab import userdata
import textwrap
import panel as pn

api_key = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)

print("OpenAI client initialized successfully.")
```

## Helper Functions

These functions:
- send prompts to the model
- maintain conversational memory
- simulate real AI assistant interactions


```
def get_completion(prompt, model="gpt-4.1", temperature=0):

    messages = [
        {"role": "user", "content": prompt}
    ]

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature
    )

    return response.choices[0].message.content


def get_completion_from_messages(messages,
                                 model="gpt-4.1",
                                 temperature=0.2):

    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature
    )

    return response.choices[0].message.content
```

## Moderation Layer

In production AI systems, moderation is essential.

This step helps:
- filter harmful prompts
- prevent abuse
- improve enterprise reliability
- support safer AI deployments


```
moderation_example = '''
We should break into the company's infrastructure
and steal sensitive pipeline credentials.
'''

response = client.moderations.create(
    input=moderation_example
)

print(response.results[0])
```

## Defining Workflow Modes

The assistant can operate in different Data Engineering modes.


```
workflow_modes = {

    "ETL": {
        "description": "Traditional Extract, Transform, Load workflow."
    },

    "ELT": {
        "description": "Extract, Load, Transform workflow."
    },

    "Medallion": {
        "description": "Bronze, Silver, and Gold layered architecture."
    },

    "Pipeline Monitoring": {
        "description": "Monitoring failed jobs and orchestration issues."
    }
}
```

```
def get_workflow_mode_details(mode_name):

    return workflow_modes.get(
        mode_name,
        {"description": "Workflow mode not found."}
    )


print(get_workflow_mode_details("Medallion"))
```

## System Prompt

This is the orchestration layer of the assistant.


```
delimiter = "####"

system_message = f'''
You are an AI Workflow Automation Assistant
specializing in Data Engineering and Analytics Systems.

The user's request will be delimited with {delimiter} characters.

Your responsibilities:
1. Understand the business problem.
2. Recommend workflow architecture.
3. Explain ETL/ELT pipeline logic.
4. Suggest scalable approaches.
5. Provide concise implementation guidance.

Return answers in a structured business-friendly format.
'''
```

## Example Business Request

```
user_message = f'''
####
My company is an e-commerce retailer.

We currently use manual Python ETL scripts,
but dashboards are delayed and workflows are becoming difficult to scale.

We want guidance on:
- improving pipeline scalability
- organizing data layers
- monitoring failed jobs
- supporting future AI analytics workflows
####
'''

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_message}
]

response = get_completion_from_messages(messages)

print(textwrap.fill(response, width=100))
```

## Conversational Context Management

The assistant continuously appends:
- user messages
- AI responses

This allows the system to maintain continuity across interactions.


```
context = [
    {"role": "system", "content": system_message}
]

context.append({
    "role": "user",
    "content": user_message
})

assistant_response = get_completion_from_messages(context)

context.append({
    "role": "assistant",
    "content": assistant_response
})

print("Conversation memory updated successfully.")
```

## Continuous Conversation Loop

This simulates how production AI assistants:
- maintain memory
- support follow-up questions
- evolve context dynamically


```
while True:

    followup = input("Ask a follow-up question (or type 'exit'): ")

    if followup.lower() == "exit":
        print("Session ended.")
        break

    context.append({
        "role": "user",
        "content": followup
    })

    response = get_completion_from_messages(context)

    context.append({
        "role": "assistant",
        "content": response
    })

    print("\nAssistant Response:\n")
    print(textwrap.fill(response, width=100))
```

## Optional Lightweight UI (Panel)

This demonstrates how backend AI logic can later connect to:
- Streamlit
- FastAPI
- dashboards
- enterprise business tools


```
pn.extension()

panels = []

context = [
    {'role': 'system', 'content': system_message}
]

inp = pn.widgets.TextInput(
    placeholder='Ask about your ETL pipeline...'
)

button = pn.widgets.Button(name="Send")

def collect_messages(event):

    prompt = inp.value
    inp.value = ''

    context.append({
        'role': 'user',
        'content': prompt
    })

    response = get_completion_from_messages(context)

    context.append({
        'role': 'assistant',
        'content': response
    })

    panels.append(
        pn.Row("User:", pn.pane.Markdown(prompt, width=600))
    )

    panels.append(
        pn.Row("Assistant:", pn.pane.Markdown(response, width=600))
    )

    return pn.Column(*panels)

interactive_conversation = pn.bind(collect_messages, button)

dashboard = pn.Column(
    inp,
    button,
    interactive_conversation
)

dashboard
```
