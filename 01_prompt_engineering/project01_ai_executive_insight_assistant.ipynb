## AI Executive Insight Assistant

### Overview
This project demonstrates applied Prompt Engineering techniques using the OpenAI API to transform raw business metrics into executive-ready insights and recommendations.

The assistant supports multiple response modes:
- Executive Summary Mode
- Analyst Mode
- Action Recommendation Mode

Built using:
- Python
- OpenAI API
- Prompt Engineering

```python
import pandas as pd
import openai
import os

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv())

openai.api_key  = os.getenv('OPENAI_API_KEY')
```

```python
from openai import OpenAI
from google.colab import userdata
import textwrap

# Retrieving the API key from Colab Secrets
api_key_value = userdata.get('OPENAI_API_KEY') # Ensure we have this "OPENAI_API_KEY" as a Secret in our Colab

client = OpenAI(api_key = api_key_value)

def get_completion(prompt, model = "gpt-3.5-turbo"):
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model = model,
        messages = messages,
        temperature = 0,
    )
    return response.choices[0].message.content
```

### Interactive Business Metrics Analyzer

```python
# 1A: Analyst Mode Selector

print("Welcome to the AI Executive Insight Assistant")

print("Choose your Professional mode:")
print("1 - Exeecutive Strategy Consultant")
print("2 - Financial Analyst")
print("3 - Marketing Analyst")
print("4 - CEO")
print("5 - Organizational Psychologist")

choice = input("\nEnter the number of your choice: ")
```

```python
# 1B: Mapping the Choice to the Role

if choice == "1":
    analyst_role = "Executive Strategy Consultant"

elif choice == "2":
    analyst_role = "Financial Analyst"

elif choice == "3":
    analyst_role = "Marketing Analyst"

elif choice == "4":
    analyst_role = "CEO"

elif choice == "5":
    analyst_role = "Organizational Psychologist"

else:
    analyst_role = "General Business Analyst"
```

```python
# 2: User Business Context

industry = input("\nEnter the industry: ") # E.g.: Retail, Healthcare,

while True:
    business_goal = input("Enter the main business goal (max 80 characters): ")
    if len(business_goal) <= 80:
        break
    else:
        print("Error: Business goal exceeds 80 characters. Please try again.")
```

```python
# 3: User KPI Inputs

revenue_change = input("\nRevenue change (%): ")
retention_change = input("Customer retention change (%): ")
marketing_change = input("Marketing spend change (%): ")
traffic_change = input("Website traffic change (%): ")
```

```python
# 4: Creating the System's Message (after the user selects his/her preferred Mode)

system_message = f"""
You are an expert {analyst_role}.
Please provide practical, data-driven business \
recommendations using modern analytical best practices. \

You are assisting a company in the {industry} industry. \


The company's primary business goal is:
{business_goal}

Your role is to:
- analyze business metrics
- identify trends and risks
- provide strategic recommendations
- explain findings clearly
- communicate professionally

Always provide actionable business insights.

Lastly, this should be your general approach to solving \
any client's problem: \

- You firstly welcome the client and thank him/her for choosing us. \
- You politely ask the client to provide you with his/her industry \
and, following that, his/her main business goal. \
- You then request the client to input his/her metrics.


"""
```

```python
# 5: Initialize Context

context = [
    {'role': 'system', 'content': system_message}
]
```

```python
# 6: Adding User Metrics into the Context

business_metrics = f"""
Revenue changed by {revenue_change}%.
Customer retention changed by {retention_change}%.
Marketing spend changed by {marketing_change}%.
Website traffic changed by {traffic_change}%.
"""

context.append({
    'role': 'user',
    'content': f"""
Here are the latest business metrics:

{business_metrics}

Please provide:
1. Executive Summary
2. Key Risks
3. Root Cause Analysis
4. Strategic Recommendations
"""
})
```

```python
# 7: Integrating the Multi-Message Helper Function, now

def get_completion_from_messages(messages, model = "gpt-4.1-mini", temperature = 0.2):
    response = client.chat.completions.create(
        model = model,
        messages = messages,
        temperature = temperature,
    )
#     print(str(response.choices[0].message))
    return response.choices[0].message.content
```

```python
# 8: Initial AI Response

response = get_completion_from_messages(context)

print("\nAI Business Analyst:\n")
print(response)
```

```python
# 9: Continuous Conversation Loop

while True:

    user_followup = input("\nAsk a follow-up question (or type 'exit'): ")

    if user_followup.lower() == "exit":
        print("Session ended.")
        break

    context.append({ # "context" = This AI's temporary conversation notebook (stored locally, sent to OpenAI's servers with each API call)
        'role': 'user',
        'content': user_followup # Adds the User's latest message to all prior Conversation history
    })

    response = get_completion_from_messages(context) # now, each time we call this "get_compl..." API, we resend the entire latest notebook to the AI (OpenAI's API servers), which then processes it to generate a response

    context.append({
        'role': 'assistant',
        'content': response # This stores the AI's response too
    })

# ^This way, the whole of the latest user-AI conversation is updated and stored by the AI to have all the latest context in mind and continually update
# its knowledge for the ongoing conversation

    print("\nAI Consultant:\n")
    print(response)
```
